```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
def split_smiles_csv(
    input_csv: str,
    train_path: str = "train.csv",
    test_path: str = "test.csv",
    test_size: float = 0.2,
    random_state: int = 42
):
    """
    Reads a CSV file with SMILES and shift_value columns, splits into train/test, and saves to files.

    Args:
        input_csv (str): Path to the original CSV file.
        train_path (str): Path to save the training CSV.
        test_path (str): Path to save the testing CSV.
        test_size (float): Proportion of the dataset to include in the test split.
        random_state (int): Random seed for reproducibility.
    """
    # Step 1: Load CSV
    df = pd.read_csv(input_csv)

    # Check required columns
    if not {'SMILES', 'shift_value'}.issubset(df.columns):
        raise ValueError("CSV must contain 'SMILES' and 'shift_value' columns.")

    # Step 2: Split the dataset
    train_df, test_df = train_test_split(
        df,
        test_size=test_size,
        random_state=random_state,
        shuffle=True
    )

    # Step 3: Save to CSV
    train_df.to_csv(train_path, index=False)
    test_df.to_csv(test_path, index=False)

    print(f"Split complete: {len(train_df)} training samples, {len(test_df)} test samples.")
    print(f"Saved train to: {train_path}")
    print(f"Saved test to:  {test_path}")

# Example usage
if __name__ == "__main__":
    split_smiles_csv("FNMR.csv",train_path="../traditional_machine_learning/data/train.csv",test_path="../traditional_machine_learning/data/test.csv")
```
